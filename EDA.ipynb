```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
datos = pd.read_csv('DF_E1.csv')
datos['Date'] = pd.to_datetime(datos['Date'])
datos.set_index('Date', inplace=True)

for col in datos.columns:
    if col != "Date":
        datos[col] = pd.to_numeric(datos[col], errors="coerce")
        
print(datos.shape)
datos.head()
```

```python
df = datos.copy()

variables = [
    "CO (ppm)", "NO (ppb)", "NO2 (ppb)", "NOX (ppb)", "O3 (ppb)", "PM10 (ug/m3)",
    "PM2.5 (ug/m3)", "PRS (mmHg)", "RAINF (mm/h)", "RH (%)",
    "SO2 (ppb)", "SR (kW/m2)", "TOUT (ºC)", "WSR (km/h)", "WDR (azimutal)"
]

# Nombres de las estaciones
estaciones = ["SURESTE", "NORESTE", "CENTRO", "NOROESTE", "SUROESTE", "NOROESTE2", "NORTE", "NORESTE2", "SURESTE2", "SUROESTE2", "SURESTE3", "SUR", "NORTE2", "NORESTE3", "NOROESTE3"]

# Generamos los nombres de columnas finales
nuevas_columnas = []
for est in estaciones:
    for var in variables:
        nuevas_columnas.append(f"{est}_{var}")

print(df.shape)
df.columns = nuevas_columnas
df.head()
```

```python
plt.figure(figsize=(10, 8))
plt.scatter(datos['SURESTE.1'], datos['SURESTE.2'])
plt.title('NO vs NO2')
plt.xlabel('NO (ppb)')
plt.ylabel('NO2 (ppb)')
plt.show()
```
